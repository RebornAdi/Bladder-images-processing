```python
import cv2
image = cv2.imread(r"C:\Users\Aditya Atul Deshmukh\Desktop\Nidhi_proj\data\processed\images\img012.jpg")
```

```python

# (x1, y1) = top-left corner
# (x2, y2) = bottom-right corner
start_point = (50, 50)
end_point = (300, 300)

# Color in BGR (Blue, Green, Red)
color = (0, 255, 0)  # Green

# Thickness (use -1 to fill rectangle)
thickness = 2

# Draw rectangle
cv2.rectangle(image, start_point, end_point, color, thickness)

# Show image
cv2.imshow("Rectangle Image", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
